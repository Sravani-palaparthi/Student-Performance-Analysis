```python
# @title Default title text

import pandas as pd

# Create student dataset
data = {
    "Name": ["Asha", "Ravi", "Priya", "Rahul",
             "Sneha", "Arjun", "Kiran", "Divya"],
    "Python": [85, 70, 95, 60, 88, 75, 65, 92],
    "Maths": [90, 65, 92, 55, 85, 80, 70, 96],
    "English": [80, 75, 88, 65, 90, 72, 68, 94]
}

df = pd.DataFrame(data)

# Display dataset
print(df)

# Calculate average marks
df["Average"] = df[["Python", "Maths", "English"]].mean(axis=1)

# Find highest scoring student
top_student = df.loc[df["Average"].idxmax()]

print("Student Performance:")
print(df)

print("\nTop Performing Student:")
print(top_student)

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.bar(df["Name"], df["Average"], color="skyblue")

plt.title("Student Performance Analysis")
plt.xlabel("Student Name")
plt.ylabel("Average Marks")

plt.show()


```
